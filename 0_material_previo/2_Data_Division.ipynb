#**5. DIVISIÓN DE DATOS**

Principio fundamental:

*Todas las transformaciones que aprenden algo de los datos deben hacerse después de dividir entre entrenamiento y prueba.*

Es decir:

1. Primero divides tu dataset → train, test (y opcionalmente val).

2. Luego ajustas (fit) tus transformaciones (feature vectors y/o embeddings) solo con el conjunto de entrenamiento.

3. Finalmente aplicas esas mismas transformaciones (ya ajustadas) a test y val.

Si no se sigue ese orden incurriremos en el problema llamado **DATA LEAKAGE**.

El **data leakage** ocurre cuando información del conjunto de prueba (datos de test) influye, aunque sea indirectamente, en la construcción del modelo o sus features.

Por ejemplo:

| Transformación                        | Si la haces antes de dividir...                   | Resultado                                                |
| ------------------------------------- | ------------------------------------------------- | -------------------------------------------------------- |
| `StandardScaler` o `MinMaxScaler`     | Aprende la media/desviación de todo el dataset    | El modelo ya “conoce” el rango del test → leakage        |
| `OneHotEncoder` con `fit` global      | Aprende todas las categorías del dataset completo | El test influye en la codificación                       |
| `Target encoding`                     | Usa promedios globales de la variable objetivo    | Mezcla info del test en el entrenamiento                 |
| `Feature selection` con `SelectKBest` | Calcula correlaciones sobre todo el dataset       | El modelo ya sabe qué features son relevantes según test |


Entonces, **SIEMPRE** se debe realizar primero la división de datos antes de hacer la construcción de los vectores de características.

```python
# Definir variables predictoras (X) y objetivo (y)
X = df.drop(columns=["categoria_delito"])  # todas las columnas menos la clase
y = df["categoria_delito"]                 # variable objetivo
```

```python
# División estratificada
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,          # 20% para prueba
    random_state=SEED,         # reproducibilidad
    stratify=y               # mantiene la proporción de categorías de delito
)
```

```python
# Confirmar tamaños y proporciones
print("División completada:")
print(f"   - Entrenamiento: {X_train.shape[0]:,} filas ({100 * X_train.shape[0] / df.shape[0]:.2f}%)")
print(f"   - Prueba:        {X_test.shape[0]:,} filas ({100 * X_test.shape[0] / df.shape[0]:.2f}%)")

# Verificación rápida de balance de clases
print("\nDistribución de clases (train vs test):")
print("Train:")
print(y_train.value_counts(normalize=True).round(3))
print("\nTest:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
# GUARDAR ENTRENAMIENTO y PRUEBA

# Combinar X e y para guardar juntos
train_df = X_train.copy()
train_df["categoria_delito"] = y_train

test_df = X_test.copy()
test_df["categoria_delito"] = y_test


# Rutas de guardado
DATA_DIR = os.path.join(BASE_DIR, "data")
os.makedirs(DATA_DIR, exist_ok=True) # Crear el directorio si no existe

train_path = os.path.join(DATA_DIR, "train_split.csv")
test_path = os.path.join(DATA_DIR, "test_split.csv")
y_train_path = os.path.join(DATA_DIR, "y_train.csv")
y_test_path = os.path.join(DATA_DIR, "y_test.csv")



# Guardar a CSV (UTF-8 con BOM para compatibilidad)
train_df.to_csv(train_path, index=False, encoding="utf-8-sig")
test_df.to_csv(test_path, index=False, encoding="utf-8-sig")
y_train.to_csv(y_train_path, index=False, encoding="utf-8-sig")
y_test.to_csv(y_test_path, index=False, encoding="utf-8-sig")


# Confirmación
print("\n Archivos guardados correctamente:")
print(f"   - Entrenamiento: {train_path}")
print(f"   - Prueba:        {test_path}")
print(f"   - y_train separado:        {y_train_path}")
print(f"   - y_test separado:         {y_test_path}")


# Verificación opcional de balance de clases
print("\nDistribución de clases (train vs test):")
print("Train:")
print(y_train.value_counts(normalize=True).round(3))
print("\nTest:")
print(y_test.value_counts(normalize=True).round(3))
```
